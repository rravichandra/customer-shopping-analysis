```python
import os
import pandas as pd
import matplotlib.pyplot as plt

path = "data/processed/clean_retail.csv"

if not os.path.exists(path):
    print("Clean file not found. Run the Step 7 cell first.")
else:
    df = pd.read_csv(path, parse_dates=["InvoiceDate"])

    # 1. "Today" = one day after the last order in the data
    snapshot = df["InvoiceDate"].max() + pd.Timedelta(days=1)

    # 2. One row per customer: Recency, Frequency, Monetary
    rfm = df.groupby("CustomerID").agg(
        Recency=("InvoiceDate", lambda x: (snapshot - x.max()).days),
        Frequency=("InvoiceNo", "nunique"),
        Monetary=("TotalPrice", "sum"),
    )

    # 3. Score each from 1 to 4 (4 is best)
    rfm["R"] = pd.qcut(rfm["Recency"], 4, labels=[4, 3, 2, 1]).astype(int)
    rfm["F"] = pd.qcut(rfm["Frequency"].rank(method="first"), 4, labels=[1, 2, 3, 4]).astype(int)
    rfm["M"] = pd.qcut(rfm["Monetary"], 4, labels=[1, 2, 3, 4]).astype(int)
    rfm["Score"] = rfm["R"] + rfm["F"] + rfm["M"]

    # 4. Name the groups
    def segment(score):
        if score >= 11: return "Champions"
        if score >= 9:  return "Loyal"
        if score >= 7:  return "Potential"
        if score >= 5:  return "At Risk"
        return "Lost"

    rfm["Segment"] = rfm["Score"].apply(segment)

    # 5. Summary table
    summary = rfm.groupby("Segment").agg(
        Customers=("Monetary", "count"),
        AvgSpend=("Monetary", "mean"),
        TotalRevenue=("Monetary", "sum"),
    ).round(0).sort_values("TotalRevenue", ascending=False)
    print(summary)

    # 6. Charts, saved for your README
    os.makedirs("images", exist_ok=True)
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    summary["Customers"].plot(kind="bar", ax=axes[0], title="Customers per Segment")
    summary["TotalRevenue"].plot(kind="bar", ax=axes[1], title="Revenue per Segment")
    plt.tight_layout()
    plt.savefig("images/rfm_segments.png", dpi=150)
    plt.show()

    rfm.to_csv("data/processed/rfm_customers.csv")
```

```python
import sqlite3
from pathlib import Path
import pandas as pd

# Locate database path
db_path = Path("data/olist.db") if Path("data/olist.db").exists() else Path("../data/olist.db")
conn = sqlite3.connect(db_path)

revenue_query = """
SELECT 
    t.product_category_name_english AS category,
    ROUND(SUM(oi.price), 2) AS total_revenue,
    COUNT(DISTINCT oi.order_id) AS total_orders,
    ROUND(SUM(oi.price) / COUNT(DISTINCT oi.order_id), 2) AS avg_order_value
FROM order_items oi
JOIN orders o ON oi.order_id = o.order_id
JOIN products p ON oi.product_id = p.product_id
JOIN product_category_name_translation t ON p.product_category_name = t.product_category_name
WHERE o.order_status = 'delivered'
GROUP BY category
ORDER BY total_revenue DESC
LIMIT 10;
"""

df_top10 = pd.read_sql(revenue_query, conn)
conn.close()

# Display formatted table
df_top10
```
